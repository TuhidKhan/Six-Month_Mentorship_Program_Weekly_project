## What was cleaned
- Column Standardizing: The original column names had a mix of spaces and slashes (e.g., race/ethnicity, math score). These were converted to lowercase snake_case standard formats (Race_ethnicity, Math_score) to avoid coding errors during selection.

- Duplicate and Missing Values Check: Checked for missing entries and duplicate rows using .isnull().sum() and .duplicated().sum(). The Kaggle dataset is natively clean, but keeping these functions in the pipeline ensures structural integrity.


## What insights were found 
- The analysis showed that students who perform well in one subject usually perform well in other subjects too. Reading and writing scores have a very strong relationship, and math scores are also connected with language skills.
- The results also showed that students with a standard lunch have higher average scores compared to students with free or reduced lunch. This suggests that students' background and available resources can affect their academic performance.
