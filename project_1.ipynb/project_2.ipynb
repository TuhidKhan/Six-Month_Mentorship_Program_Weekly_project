### **Dataset Metadata**

This table explains what each column in our dataset represents in simple terms:

| Column Name | Description |
| :--- | :--- |
| **Row ID** | A unique number for every single row in the dataset. |
| **Order ID** | A unique tracking number given to each customer's order. |
| **Order Date** | The date when the customer purchased the items. |
| **Ship Date** | The date when the items were shipped out. |
| **Ship Mode** | The delivery method chosen by the customer (e.g., First Class, Standard). |
| **Customer ID** | A unique ID number used to identify each specific customer. |
| **Customer Name** | The full name of the customer. |
| **Segment** | The group or category the customer belongs to (Consumer, Corporate, or Home Office). |
| **Country** | The country where the customer lives. |
| **City** | The city where the customer lives. |
| **State** | The state where the customer lives. |
| **Postal Code** | The zip code or postal code of the customer's delivery address. |
| **Region** | The geographic area where the customer is located (like West, East, South, Central). |
| **Product ID** | A unique identification number for each specific item sold. |
| **Category** | The broad main group the item belongs to (like Technology, Furniture, Office Supplies). |
| **Sub-Category** | The smaller, more specific group the item belongs to (like Chairs, Tables, Phones). |
| **Product Name** | The actual name and model of the product. |
| **Sales** | The total amount of money paid for the product. |
| **Quantity** | The total number of items bought in that order. |
| **Discount** | The percentage price deduction given on the purchase. |
| **Profit** | The amount of money the company made (or lost, if negative) on the order. |

```python
#Now we will see the correletion of the dateset and will plot it using heatmap



plt.figure(figsize=(16, 6))


sns.heatmap(
    df.corr(numeric_only=True), 
    annot=True,          
    cmap="coolwarm",     
    fmt=".2f",          
    linewidths=0.5,      
    vmin=-1, vmax=1,     
    cbar_kws={"shrink": 0.8} 
)


plt.title("Correlation Matrix Heatmap", fontsize=16, fontweight='bold', pad=15)
plt.tight_layout() # Prevents any text on the edges from getting cut off
plt.show()
```

### **Description of the Correlation Heatmap**

This chart shows how different numbers in our data affect each other. The biggest finding is that discounts and profit ratios have a strong negative connection (-0.86). This simply means that giving big discounts badly hurts our profit percentage. On the other hand, sales and profit have a positive connection (0.48), which proves that making more sales generally brings in more money. Other things, like the number of items sold (quantity), barely affect profit at all. Also, extra data like row IDs, zip codes, months, and years have scores near zero, meaning they don't change business performance at all. Moving forward, our strategy should be to stop giving big discounts and focus on selling products at full price to protect our profits.

```python
#Identify loss-making products
loss_df = df[df["Profit"] < 0]
top_losses = loss_df.groupby("Sub-Category")["Profit"].sum().reset_index()

# 2. Sort so the biggest loss is at the top/most visible
top_losses = top_losses.sort_values(by="Profit", ascending=True).head(10)

# 3. Plot simple and clean horizontal bar chart
fig = px.bar(
    top_losses,
    x="Profit",
    y="Sub-Category",
    orientation="h",
    title="Top Loss-Making Sub-Categories",
    color_discrete_sequence=["#d9534f"], # Clean muted red for losses
    labels={"Profit": "Total Loss (USD)", "Sub-Category": ""}
)

# Clean minimalist layout
fig.update_layout(
    template="plotly_white",
    xaxis=dict(showgrid=True, gridcolor="LightGray"),
    yaxis={'categoryorder':'total descending'} # Keeps worst loss at the top
)

fig.show()
```

### Key Insights: Top Loss-Making Sub-Categories

- **Binders**
  - **Status:** Highest Loss Contributor
  - **Impact:** Accounts for the largest deficit, driving down profits by approximately **-\$38,500**.

- **Tables & Machines**
  - **Status:** Significant Loss Drivers
  - **Impact:** Both are major pain points for the business, with **Tables** generating a total loss of roughly **-\$32,500** and **Machines** following closely behind at around **-\$30,000**.

- **Furniture & Equipment Clustered Losses**
  - **Status:** Moderate-to-High Losses
  - **Impact:** **Bookcases** (~ -\$12,000) and **Chairs** (~ -\$10,000) represent a combined furniture-related deficit that highlights systemic pricing or discounting issues within that category.

- **Minor Underperformers**
  - **Status:** Low-Tier Losses
  - **Impact:** **Supplies** is the least impactful offender on this list, safely maintaining its negative footprint under **-\$3,000**.

```python
# now will plot Monthly and yearly trends using plotly
df_line = df.groupby('Order Date')[['Sales', 'Profit']].sum().reset_index()
fig = px.line(df_line, x='Order Date', y='Profit', title='Monthly and Yearly Trends Over Time')
fig.update_xaxes(rangeslider_visible=True) # Adds a timeline slider at the bottom
fig.show()
```



#### **What the Graph Shows (Findings)**
* **Normal Days:** Most of the time, the business makes a steady, small profit every day.
* **Wild Changes:** In 2016 and 2017, the profit lines start jumping up and down much more than before.
* **Super Good Days:** There are a few days where profit went very high (over 5,000 in one day), especially in mid-2016 and early 2017.
* **Bad Days:** There are also a few days where the business lost a lot of money. For example, on **July 28, 2016**, the business lost **$1,307.07**.

---

#### **What We Should Do Next**
* **Copy the Good Days:** Find out what sold so well on the highest profit days and try to do it again.
* **Fix the Bad Days:** Check the days with big losses to see why we lost money (like too many returns or bad discounts) so we can stop it from happening again.

```python
# now we will plot scatter plot for Discount vs Profit relationship 

fig = px.scatter(df, x="Discount", y="Profit", title="Discount vs Profit Relationship")
fig.show()



```

### **Discount vs Profit Analysis**

 **What the Graph Shows (Findings)**
* **No Discounts = High Profit:** When the discount is `0` (on the far left), almost all the dots are high above the `0` line. This is where the business makes its biggest profits.
* **Small Discounts are Okay:** At `0.1` (10%) and `0.2` (20%) discounts, the business still makes a decent profit, though the numbers are lower than zero-discount sales.
* **Big Discounts Cause Big Losses:** As you look further to the right (`0.3` to `0.8`), the dots fall below the `0` line. When discounts hit **50% (`0.5`) or higher**, the business almost never makes a profit and starts losing a lot of money. 

---

 **What We Should Do Next**
* **Stop High Discounts:** Avoid giving discounts higher than 20% because they destroy our profit margins.
* **Keep Doing What Works:** Focus sales efforts on zero-discount or low-discount offers, as these keep the business profitable.
