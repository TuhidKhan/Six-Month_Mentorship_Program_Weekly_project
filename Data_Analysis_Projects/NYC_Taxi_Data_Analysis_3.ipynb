# NYC Taxi Trip Duration Dataset

## About the Dataset
This dataset is based on the 2016 NYC Yellow Cab trip record data published by the New York City Taxi and Limousine Commission (TLC). It contains key details about individual taxi rides in New York City, including pickup and dropoff times, coordinates, passenger counts, and overall trip durations. 

Primary objectives typically include:
- Performing Exploratory Data Analysis (EDA) on urban mobility patterns.
- Predicting the total duration of a ride (`trip_duration`) based on spatial and temporal features.

### 15.9 min
We calculate the average trip duration in minutes to understand typical travel times and spot unusually long or short trips in the dataset.

```python

def haversine_distance(lat1, lon1, lat2, lon2):
    R = 6371  # Earth radius in kilometers

    lat1 = np.radians(lat1)
    lon1 = np.radians(lon1)
    lat2 = np.radians(lat2)
    lon2 = np.radians(lon2)

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = np.sin(dlat / 2) ** 2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2) ** 2
    c = 2 * np.arcsin(np.sqrt(a))

    return R * c

df['trip_distance'] = haversine_distance(
    df['pickup_latitude'],
    df['pickup_longitude'],
    df['dropoff_latitude'],
    df['dropoff_longitude']
)
df.head()
```

# FEATURE SCALING / NORMALIZATION 
- Apply Min-Max scaling 
- Apply Standardization 
- Compare both approaches

```python

from sklearn.preprocessing import MinMaxScaler, StandardScaler

cols_to_scale = ["trip_duration", "trip_distance", "passenger_count", "hour"]

# Min-Max Scaling
minmax_scaler = MinMaxScaler()
df_minmax = df.copy()
df_minmax[cols_to_scale] = minmax_scaler.fit_transform(df[cols_to_scale])

# Standardization
standard_scaler = StandardScaler()
df_standard = df.copy()
df_standard[cols_to_scale] = standard_scaler.fit_transform(df[cols_to_scale])

# Compare
comparison = pd.DataFrame({
    "Original_mean": df[cols_to_scale].mean(),
    "Original_std": df[cols_to_scale].std(),
    "MinMax_mean": df_minmax[cols_to_scale].mean(),
    "MinMax_std": df_minmax[cols_to_scale].std(),
    "Standard_mean": df_standard[cols_to_scale].mean(),
    "Standard_std": df_standard[cols_to_scale].std(),
})

comparison
```

### Comparison: Min-Max vs Standardization

- **Min-Max scaling** squeezed all values between 0 and 1,  but the mean/std still differ per column depending on outliers (e.g. `trip_duration` mean is almost 0 due to extreme long trips).
  
- **Standardization** gave every column a mean ≈ 0 and std = 1, making it more consistent and reliable — this makes it the better choice for this dataset.

# 5. FEATURE ENCODING 
- Encode categorical features (e.g. passenger count categories) 
- Use label encoding and one-hot encoding 

```python
df['passenger_count'].value_counts()
```

```python

from sklearn.preprocessing import LabelEncoder

# 1. Map numerical counts into categorical bins
def categorize_passengers(count):
    if count == 1:
        return 'Single'
    elif count <= 4:
        return 'Small Group'
    else:
        return 'Large Group'

df['passenger_category'] = df['passenger_count'].apply(categorize_passengers)

# 2. Apply Label Encoding
le = LabelEncoder()
df['passenger_category_label'] = le.fit_transform(df['passenger_category'])

# Preview results
df[['passenger_count', 'passenger_category', 'passenger_category_label']].head()
```

### Step 1: Feature Binning & Label Encoding

* **`categorize_passengers()` Function**: Groups the raw continuous numerical passenger counts into three distinct categories:
  * **`Single`**: 1 passenger
  * **`Small Group`**: 2 to 4 passengers
  * **`Large Group`**: 5+ passengers
* **`passenger_category`**: Stores the readable text category (`Single`, `Small Group`, or `Large Group`).
* **`passenger_category_label`**: Converts the string categories into machine-readable integer codes using `LabelEncoder()`:
  * **`0`**: Large Group
  * **`1`**: Single
  * **`2`**: Small Group

```python
# Apply One-Hot Encoding on the categorical column
df_onehot = pd.get_dummies(df, columns=['passenger_category'], prefix='passenger', dtype=int)

# Preview the newly created binary dummy columns
df_onehot.filter(like='passenger').head()
```

### Feature Encoding 

* **`passenger_count` (`1`)**: The original raw feature value indicating a single passenger.
* **`passenger_category_label` (`1`)**: Result of **Label Encoding**, where categories are mapped to integers based on alphabetical order (`0`: Large Group, `1`: Single, `2`: Small Group).
* **`passenger_Large Group` (`0`)**: **One-Hot Encoding** flag — `0` means False (not a large group).
* **`passenger_Single` (`1`)**: **One-Hot Encoding** flag — `1` means True (is a single passenger).
* **`passenger_Small Group` (`0`)**: **One-Hot Encoding** flag — `0` means False (not a small group).

```python
def time_bucket(h):
    if 5 <= h < 12: return "Morning"
    elif 12 <= h < 17: return "Afternoon"
    elif 17 <= h < 21: return "Evening"
    else: return "Night"

df["time_of_day"] = df["hour"].apply(time_bucket)

df["distance_category"] = pd.cut(
    df["trip_distance"],
    bins=[-0.01, 2, 6, float("inf")],   # -0.01 so 0-distance trips are included
    labels=["Short", "Medium", "Long"]
)

# Safety net: drop any remaining NaNs (e.g. negative distances, bad data)
df_sunburst = df.dropna(subset=["passenger_category_label", "time_of_day", "distance_category"])

fig3 = px.sunburst(
    df_sunburst,
    path=["passenger_category_label", "time_of_day", "distance_category"],
    title="Passenger → Time of Day → Distance Category"
)
fig3.show()


 
 
```

### Interpretation: Passenger → Time of Day → Distance Category

- The sunburst shows how trips break down by **passenger category** (inner ring), then **time of day** (middle ring), then **trip distance** (outer ring).
- Ring size = number of trips — bigger slice means more trips fall in that group.
- **Inner ring:** **Single passengers (1)** make up the largest share of trips, followed by **small groups (0)**, while **large groups (2)** account for the smallest share.
- **Middle ring:** within each passenger category, trips are fairly spread across Morning, Afternoon, Evening, and Night, with no single time slot dominating heavily.
- **Outer ring:** **Short-distance trips are the most common** across almost every passenger–time combination, followed by Medium, with Long trips making up the smallest share.
- **Takeaway:** Most taxi rides are taken by single passengers on short trips, regardless of the time of day.
