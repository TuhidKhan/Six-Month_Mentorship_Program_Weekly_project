```python
# import libraries - Part 1 (time series)
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from statsmodels.tsa.stattools import adfuller
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.statespace.sarimax import SARIMAX
from sklearn.metrics import mean_absolute_error, mean_squared_error

import warnings
warnings.filterwarnings('ignore')
```

# PART 1: Time Series Forecasting (Delhi Climate)

## 1. Data Understanding

```python
# load the dataset
train_df = pd.read_csv('DailyDelhiClimateTrain.csv')
test_df = pd.read_csv('DailyDelhiClimateTest.csv')
train_df.head()
```

```python
# check shape and info
print(train_df.shape, test_df.shape)
train_df.info()
```

```python
# convert date column to datetime and set as index
train_df['date'] = pd.to_datetime(train_df['date'])
test_df['date'] = pd.to_datetime(test_df['date'])

train_df.set_index('date', inplace=True)
test_df.set_index('date', inplace=True)

print('Train date range:', train_df.index.min(), 'to', train_df.index.max())
print('Test date range:', test_df.index.min(), 'to', test_df.index.max())
```

```python
train_df.describe()
```

- minimum Temperature: 6
- maximum Temperature: 38.7

```python
# decompose the time series into trend, seasonal and residual components
from statsmodels.tsa.seasonal import seasonal_decompose
decomposition = seasonal_decompose(train_df['meantemp'], model='additive', period=365)
decomposition.plot()
```

```python
# make a plot on plotly
import plotly.express as px

fig = px.line(
    train_df.reset_index(),
    x='date',
    y=['meantemp'],
    title='Delhi Climate Time Series'
)
fig.show()
```

## 2. Data Preprocessing

### Checking Missing values

```python
# check for missing values
print(train_df.isnull().sum())
print(test_df.isnull().sum())
```

- There is no missing values in the dataset

### Check stationarity

```python
# apply ADF test on meantemp
result = adfuller(train_df['meantemp'])

print('ADF Statistic:', result[0])
print('p-value:', result[1])

if result[1] < 0.05:
    print('Result: Series is stationary')
else:
    print('Result: Series is NOT stationary - differencing needed')
```

```python
# apply differencing since temperature has a clear yearly seasonal cycle (not stationary)
temp_diff = train_df['meantemp'].diff().dropna()

result_diff = adfuller(temp_diff)
print('ADF Statistic (differenced):', result_diff[0])
print('p-value (differenced):', result_diff[1])

plt.figure(figsize=(12, 4))
plt.plot(temp_diff)
plt.title('Differenced Mean Temperature')
plt.show()
```

## 3. ARIMA Model

```python
# build ARIMA model on training data
arima_model = ARIMA(train_df['meantemp'], order=(3, 1, 2))
arima_fit = arima_model.fit()

# forecast for the test period
arima_forecast = arima_fit.forecast(steps=len(test_df))

print(arima_fit.summary())
```

```python
# evaluate ARIMA
print('MAE:', mean_absolute_error(test_df['meantemp'], arima_forecast))
print('RMSE:', np.sqrt(mean_squared_error(test_df['meantemp'], arima_forecast)))
```

## 4. SARIMA / SARIMAX
Temperature has a strong yearly seasonal cycle - SARIMA's seasonal terms (period=365 for daily data, approximated with 12 for monthly-like cycles here) should capture this better than plain ARIMA.

```python
# build SARIMA model with yearly seasonality
sarima_model = SARIMAX(train_df['meantemp'], order=(3, 1, 2), seasonal_order=(1, 1, 1, 12))
sarima_fit = sarima_model.fit(disp=False)

# forecast for the test period
sarima_forecast = sarima_fit.forecast(steps=len(test_df))

print(sarima_fit.summary())
```

### Compare ARIMA vs SARIMA

```python
# compare ARIMA vs SARIMA visually
plt.figure(figsize=(12, 5))
plt.plot(test_df.index, test_df['meantemp'], label='Actual', marker='o')
plt.plot(test_df.index, arima_forecast, label='ARIMA Forecast', linestyle='--')
plt.plot(test_df.index, sarima_forecast, label='SARIMA Forecast', linestyle='--')
plt.title('ARIMA vs SARIMA Forecast - Delhi Temperature')
plt.legend()
plt.show()
```

```python
# evaluate SARIMA and compare with ARIMA
print('SARIMA MAE:', mean_absolute_error(test_df['meantemp'], sarima_forecast))
print('SARIMA RMSE:', np.sqrt(mean_squared_error(test_df['meantemp'], sarima_forecast)))
```

## 5. Prophet Model

```python
from prophet import Prophet

# prepare data in Prophet's required format
prophet_train = train_df.reset_index()[['date', 'meantemp']]
prophet_train.columns = ['ds', 'y'] # ds = date, y = target

prophet_train.head()
```

```python
from prophet import Prophet

# call and fit the model
prophet_model = Prophet(
    yearly_seasonality=True,
    daily_seasonality=False
)
prophet_model.fit(prophet_train)

# create future dataframe covering the test period
future = prophet_model.make_future_dataframe(periods=len(test_df), freq='D')

# forecast
prophet_forecast = prophet_model.predict(future)

# plot the forecast
fig1 = prophet_model.plot(prophet_forecast)
plt.title('Prophet Forecast - Delhi Temperature')
plt.show()

fig2 = prophet_model.plot_components(prophet_forecast)
plt.show()

#plot the same plot in plotly
# Use the existing Prophet forecast and training data
forecast = prophet_forecast
df = prophet_train

fig = px.line(
    forecast,
    x="ds",
    y="yhat",
    title="Prophet Forecast",
    labels={"ds": "Date", "yhat": "Predicted temperature"}
)

fig.add_scatter(
    x=forecast["ds"],
    y=forecast["yhat_lower"],
    mode="lines",
    line=dict(width=0),
    showlegend=False,
    name="Lower bound"
)

fig.add_scatter(
    x=forecast["ds"],
    y=forecast["yhat_upper"],
    mode="lines",
    fill="tonexty",
    line=dict(width=0),
    name="Uncertainty interval"
)

fig.add_scatter(
    x=df["ds"],
    y=df["y"],
    mode="markers",
    name="Actual"
)

fig.show()
```

```python
# evaluate prophet on the test period
prophet_test_pred = prophet_forecast.set_index('ds').loc[test_df.index, 'yhat']

prophet_mae = mean_absolute_error(test_df['meantemp'].values, prophet_test_pred.values)
prophet_rmse = np.sqrt(mean_squared_error(test_df['meantemp'].values, prophet_test_pred.values))
print('Prophet MAE:', prophet_mae)
print('Prophet RMSE:', prophet_rmse)
```

## 6. Model Evaluation

```python
# compare all models
comparison = pd.DataFrame({
    'Model': ['ARIMA', 'SARIMA', 'Prophet'],
    'MAE': [
        mean_absolute_error(test_df['meantemp'], arima_forecast),
        mean_absolute_error(test_df['meantemp'], sarima_forecast),
        prophet_mae
    ],
    'RMSE': [
        np.sqrt(mean_squared_error(test_df['meantemp'], arima_forecast)),
        np.sqrt(mean_squared_error(test_df['meantemp'], sarima_forecast)),
        prophet_rmse
    ]
})
comparison.sort_values('RMSE')
```

# PART 2: Model Improvement

```python
# check for overfitting by comparing train vs test performance for SARIMA
train_pred = sarima_fit.fittedvalues
train_mae = mean_absolute_error(train_df['meantemp'], train_pred)
test_mae = mean_absolute_error(test_df['meantemp'], sarima_forecast)

print('SARIMA Train MAE:', train_mae)
print('SARIMA Test MAE:', test_mae)
print('Gap:', test_mae - train_mae)
```

### Improve model performance
Try different ARIMA/SARIMA orders and compare AIC (lower AIC = better balance of fit vs complexity) to find better parameters than the initial guess.

```python
# try a few different SARIMA parameter combinations and compare AIC
orders_to_try = [
    ((1,1,1), (1,1,1,12)),
    ((2,1,2), (1,1,1,12)),
    ((1,1,0), (0,1,1,12)),
]

results = []
for order, seasonal_order in orders_to_try:
    m = SARIMAX(train_df['meantemp'], order=order, seasonal_order=seasonal_order)
    f = m.fit(disp=False)
    results.append({'order': order, 'seasonal_order': seasonal_order, 'AIC': f.aic})

results_df = pd.DataFrame(results).sort_values('AIC')
results_df

```

# PART 3: Transfer Learning (Intel Image Classification)
Dataset: 6 scene classes (buildings, forest, glacier, mountain, sea, street) - ~14,000 training images, ~3,000 test images.

```python
# import libraries 
import os
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.applications import MobileNetV2 # import MobileNetV2 is a pre-trained model
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import EarlyStopping

tf.random.set_seed(42)
np.random.seed(42)
```

## 1. Data Preprocessing

```python
# paths to the image folders
train_dir = 'seg_train/seg_train'
test_dir = 'seg_test/seg_test'

# check classes and image counts
classes = os.listdir(train_dir)
print('Classes:', classes)

# check image counts
for c in classes:
    train_count = len(os.listdir(os.path.join(train_dir, c)))
    test_count = len(os.listdir(os.path.join(test_dir, c)))
    print(f'{c}: {train_count} train, {test_count} test')
```

```python
# set image size and batch size - MobileNetV2 default input size is 224x224
IMG_SIZE = 224
BATCH_SIZE = 32

# create ImageDataGenerator - resize, normalize, and split a validation set from training data
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=15,
    horizontal_flip=True,
    zoom_range=0.1,
    validation_split=0.15
)

test_datagen = ImageDataGenerator(rescale=1./255)

# load training images
train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='training',
    seed=42
)

# load validation images
val_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='validation',
    seed=42
)

# load test images
test_generator = test_datagen.flow_from_directory(
    test_dir,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=False
)

print('Class indices:', train_generator.class_indices)
```

```python
# visualize a few sample images
images, labels = next(train_generator)
class_names = list(train_generator.class_indices.keys())

plt.figure(figsize=(12, 6))
for i in range(8):
    plt.subplot(2, 4, i+1)
    plt.imshow(images[i])
    plt.title(class_names[np.argmax(labels[i])])
    plt.axis('off')
plt.tight_layout()
plt.show()
```

## 2. Transfer Learning - MobileNetV2
Instead of training a CNN from scratch, we use MobileNetV2 - a model already pre-trained on ImageNet (millions of images, 1000 classes). We freeze its learned layers (so we don't lose that knowledge) and only train new layers on top for our 6 specific classes.

```python
# load MobileNetV2 pre-trained on ImageNet, without its original top classification layer
base_model = MobileNetV2(
    input_shape=(IMG_SIZE, IMG_SIZE, 3),
    include_top=False, # include_top=False: don't include the top classification layer
    weights='imagenet' # weights='imagenet': load weights pre-trained on ImageNet
)

# freeze the base model - its pre-trained weights won't be updated during training
base_model.trainable = False #base_model.trainable = False: freeze the base model means its pre-trained weights won't be updated during training

print('Number of layers in base model:', len(base_model.layers))
```

## 3. Model Building
Add our own custom classification layers on top of the frozen MobileNetV2 base.

```python
# build the full model - frozen base + custom classification head
model = keras.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),   # reduces each feature map to a single number
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.3),
    layers.Dense(6, activation='softmax')   # 6 classes, softmax gives probability per class
])

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

## 4. Model Training

```python
# early stopping - stop training if validation loss doesn't improve
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)

# train the model - only the custom top layers actually get updated since base is frozen
history = model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=10,
    callbacks=[early_stop],
    verbose=1
)
```

## 5. Evaluation

```python
# evaluate on the test set
test_loss, test_accuracy = model.evaluate(test_generator, verbose=0)
print('Test Loss:', test_loss)
print('Test Accuracy:', test_accuracy)
```
