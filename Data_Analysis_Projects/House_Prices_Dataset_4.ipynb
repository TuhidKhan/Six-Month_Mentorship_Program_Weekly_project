# 1. BASIC MATH APPLICATION 
- Calculate percentages (price change, ratios) 
- Work with fractions and proportions from data 

```python
avg_price = df['price'].mean()
df['price_pct_vs_avg'] = ((df['price'] - avg_price) / avg_price) * 100
df[['price', 'price_pct_vs_avg']]
```

```python
# Ratio: living space vs lot size


df['living_to_lot_ratio'] = df['sqft_living'] / df['sqft_lot']
df[['sqft_living', 'sqft_lot', 'living_to_lot_ratio']]
```

```python
# Proportion: basement space out of total living space

df['basement_proportion'] = df['sqft_basement'] / df['sqft_living']
df[['sqft_basement', 'sqft_living', 'basement_proportion']]
```

```python
# % of homes that are waterfront

waterfront_pct = (df['waterfront'].sum() / len(df)) * 100
waterfront_pct
```


- price_pct_vs_avg: how far each house's price is from the average price (%)
- living_to_lot_ratio: how much of the lot is used by living space
- basement_proportion: fraction of living space that is basement
- waterfront_pct: % of total homes that are waterfront properties

# 2. ALGEBRA (EQUATIONS) 
- Form linear equations using dataset features 
- EXAMPLE: \
Price = a*(sqft) + b*(bedrooms) + c 

```python
# Fit the equation using Linear Regression

from sklearn.linear_model import LinearRegression

X = df[['sqft_living', 'bedrooms']]
y = df['price']

model = LinearRegression()
model.fit(X, y)

a, b = model.coef_
c = model.intercept_

a, b, c
```

## Equation
Price = a * sqft_living + b * bedrooms + c

(a, b, c values printed above from the model)

```python
# Use the equation to predict price for a sample house
df['predicted_price'] = a * df['sqft_living'] + b * df['bedrooms'] + c
df[['price', 'predicted_price']]
```

```python
# Check how close predictions are (error)
df['prediction_error'] = df['price'] - df['predicted_price']
df[['price', 'predicted_price', 'prediction_error']]
```


- a: price increase per additional sqft
- b: price increase per additional bedroom
- c: base price when sqft and bedrooms are 0 (intercept)
- prediction_error: difference between actual and predicted price

# 5. SOLVE SYSTEM OF EQUATIONS 
- SOLVE AX = B USING: 
- Inverse method 
- NumPy functions 

```python
# Create a square matrix A and vector B

A = df[['sqft_living', 'bedrooms']].to_numpy()[:2]
B = df['price'].to_numpy()[:2]
A, B
```

```python
# Solve using Inverse method
A_inv = np.linalg.inv(A)
X_inverse = A_inv @ B
X_inverse
```

```python
# Solve using NumPy's built-in function
X_solve = np.linalg.solve(A, B)
X_solve
```


- A: square matrix (2 features, 2 houses) — needs to be square to solve directly
- B: target values (price) for those 2 houses
- X_inverse: solution found by manually inverting A and multiplying by B
- X_solve: solution found using NumPy's solve() function (faster, more stable)
- Both methods should give the same answer

# 7. EIGENVALUES & EIGENVECTORS 
- Calculate eigenvalues 
- Understand meaning in data 

```python
# Select features and compute covariance matrix

features = df[['sqft_living', 'bedrooms', 'bathrooms']]
cov_matrix = np.cov(features.T)
cov_matrix
```

```python
# Calculate eigenvalues and eigenvectors


eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)
eigenvalues, eigenvectors
```


- cov_matrix: shows how features vary together (covariance between sqft_living, bedrooms, bathrooms)
- eigenvalues: show how much "spread" (variance) exists along each eigenvector direction
- eigenvectors: show the direction of that spread
- Largest eigenvalue = direction with most variation in the data (most important pattern)

# 8. SVD (SINGULAR VALUE DECOMPOSITION) 
- Apply SVD using NumPy 
- Interpret results 

```python
# Select features and apply SVD

features_matrix = df[['sqft_living', 'bedrooms', 'bathrooms']].to_numpy()[:100]
U, S, VT = np.linalg.svd(features_matrix, full_matrices=False)
U, S, VT
```

```python
# Check shapes of each part

U.shape, S.shape, VT.shape
```

```python
# Reconstruct the original matrix from U, S, VT

reconstructed = U @ np.diag(S) @ VT
reconstructed

```


- U: shows relationships between rows (houses)
- S: singular values — importance/strength of each pattern (largest first)
- VT: shows relationships between columns (features)
- reconstructed: rebuilding original data from U, S, VT — should match features_matrix
